```python
# PLOTTING THE DISTRIBUTION OF STUDY HOURS USING A HISTOGRAM:

# Q) How many hours do students typically study?

plt.figure(figsize=(10,6))

sns.histplot(df["study_hours_per_day"], bins=10)

plt.title("Distribution of Study Hours")
plt.xlabel("Study Hours")
plt.ylabel("Number of Students")

plt.show()

# ANALYSIS:
# 4–5 hours appears to be the most common study-hour range in this dataset.
# The number of students studying for 1 - 2 hours is lower that that of the students studying for 3 - 6 hours
# Study hours are concentrated around 3–6 hours per day, with the highest frequency occurring around 4–5 hours.
```

```python
# HISTOGRAM ANALYSIS:

# X-axis → study hours
# Y-axis → number of students
# Each bar → how many students fall within that study-hour range
```

```python
# PLOTTING THE DISTRIBUTION OF SLEEP HOURS USING HISTOGRAM:
# Q) How many hours do students typically sleep?

plt.figure(figsize=(12,6))
sns.histplot(df["sleep_hours"], bins=10)
plt.title("Distribution of Sleep hours")
plt.xlabel("Sleep hours")
plt.ylabel("No. of Students")
plt.show()

# ANALYSIS:
# The sleep hours are concentrated around 6 - 8 hours per day, with the highest number of stidents sleeping for about 7 hours each day.
# The number of students with 4 - 5 hours of sleep is lower than the number of students with 7 - 8 hours of sleep
```

```python
# PLOTTNG THE DISTRIBUTION OF STUDENT ATTENDANCE USING HISTOGRAM:
# Q) How is student attendance distributed?
plt.figure(figsize=(12,6))
sns.histplot(df["attendance_percentage"], bins =10)

plt.title("Distribution of Student Attendance")
plt.xlabel("Attendance (%)")
plt.ylabel("No. of Students")
plt.show()
```

```python
# PLOTTING THE DISTRIBUTION OF EXAM SCORES USING HISTOGRAM:
# Q) How are student scores distributed?
plt.figure(figsize=(12, 6))
sns.histplot(df["exam_score"], bins= 10)

plt.title("Student exam scores Distribution")
plt.xlabel("Exam Score")
plt.ylabel("No. of Students")
plt.show()

# ANALYSIS:
# the number of students increases with increase in exam scores.
# Around 13000 students having around a 90 exam score with a significant increase to around 38000 number of students having a score of 100.
```

```python
# ANALYSING THE RELATIONSHIP BETWEEN STUDY HOURS AND EXAM SCORES USING A SCATTER PLOT:
# Q) Do students who study more tend to achieve higher scores?
plt.figure(figsize=(12, 6))

sns.scatterplot(
    x= "study_hours_per_day",
    y = "exam_score",
    data= df
)

plt.title("Study Hours VS Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.show()
```

```python
# CALCULATING THE CORRELATION BETWEEN STUDY HOURS AND EXAM SCORES:
# Q) How strong is the relationship between study hours and exam scores?
df["study_hours_per_day"].corr(df["exam_score"])

# No linear relationship between study hours and exam scores
```

```python
# ANALYSING THE RELATIONSHIP BETWEEN SLEEP HOURS AND EXAM SCORE:
# Q) Do students who sleep more tend to have higher exam scores?
plt.figure(figsize=(12, 6))
sns.scatterplot(
    x="sleep_hours", 
    y="exam_score",
    data=df
)

plt.title("Sleep Hours VS Exam Score")
plt.xlabel("Sleep Hours")
plt.ylabel("Exam Score")
plt.show()
```

```python
# CALCULATING THE CORRELATION BETWEEN SLEEP HOURS AND EXAM SCORE:
# Q) How strong is the relationship between sleep hours and exam scores?
df["sleep_hours"].corr(df["exam_score"])

# No linear relationship between sleep hours and exam scores
```

```python
# ANALYSING THE RELATIONSHIP BETWEEN ATTENDANCE AND EXAM SCORES:
# Q) Do students with higher attendance tend to have higher exam scores?
plt.figure(figsize=(12,6))
sns.scatterplot(
    x= ("attendance_percentage"),
    y=("exam_score"),
    data=df 
)

plt.title("Attendance VS Exam score")
plt.xlabel("Attendance (%)")
plt.ylabel("Exam score")
plt.show()
```

```python
# CALCULATING THE CORRELATION BETWEEN ATTENDANCE AND EXAM SCORE:
df["attendance_percentage"].corr(df["exam_score"])

# No linear relationship between attendance precentage and exam score
```
