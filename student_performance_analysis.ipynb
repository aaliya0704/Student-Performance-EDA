```python
# PLOTTING THE DISTRIBUTION OF STUDY HOURS USING A HISTOGRAM:

# Q) How many hours do students typically study?

plt.figure(figsize=(10,6))

sns.histplot(df["study_hours_per_day"], bins=10)

plt.title("Distribution of Study Hours")
plt.xlabel("Study Hours")
plt.ylabel("Number of Students")

plt.show()

# ANALYSIS:
# Around 19000 students study for 4 hours
```

```python
# HISTOGRAM ANALYSIS:

# X-axis → study hours
# Y-axis → number of students
# Each bar → how many students fall within that study-hour range
```

```python
# PLOTTING THE DISTRIBUTION OF SLEEP HOURS USING HISTOGRAM:
# Q) How many hours do students typically sleep?

plt.figure(figsize=(12,6))
sns.histplot(df["sleep_hours"], bins=10)
plt.title("Distribution of Sleep hours")
plt.xlabel("Sleep hours")
plt.ylabel("No. of Students")
plt.show()
```

```python
# PLOTTNG THE DISTRIBUTION OF STUDENT ATTENDANCE USING HISTOGRAM:
# Q) How is student attendance distributed?
plt.figure(figsize=(12,6))
sns.histplot(df["attendance_percentage"], bins =10)

plt.title("Distribution of Student Attendance")
plt.xlabel("Attendance (%)")
plt.ylabel("No. of Students")
plt.show()
```

```python
# PLOTTING THE DISTRIBUTION OF EXAM SCORES USING HISTOGRAM:
# Q) How are student scores distributed?
plt.figure(figsize=(12, 6))
sns.histplot(df["exam_score"], bins= 10)

plt.title("Student exam scores Distribution")
plt.xlabel("Exam Score")
plt.ylabel("No. of Students")
plt.show()
```
